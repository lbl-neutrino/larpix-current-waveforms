```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
```

```python
Path('plots').mkdir(exist_ok=True)
```

```python
for fname in sorted(Path('waveforms').glob('*.txt')):
    arr = np.loadtxt(fname)
    plt.figure()
    plt.plot(arr)
    plt.title(fname.stem)
    plt.xlabel(R'Tick [0.1 $\mu$s]')
    plt.ylabel(R'Pixel current [e$^-$/$\mu$s]')
    plt.savefig(f'plots/{fname.stem}.png')
```

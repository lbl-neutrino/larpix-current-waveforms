```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.style.use('dark_background')
```

```python
Path('plots/waveforms').mkdir(parents=True, exist_ok=True)
```

```python
# for fname in sorted(Path('waveforms').glob('blip_x10cm_z4.25p*.txt')):
for fname in sorted(Path('waveforms').glob('blip_*.txt')):
    arr = np.loadtxt(fname)
    plt.figure()
    plt.plot(arr)
    plt.title(fname.stem)
    plt.xlabel(R'Tick [0.1 $\mu$s]')
    plt.ylabel(R'Pixel current [e$^-$/$\mu$s]')
    plt.tight_layout()
    plt.savefig(f'plots/waveforms/{fname.stem}.png')
```
